- In Keras 3.0, the backend must be set before importing Keras.
- Also, the backend cannot be changed after Keras has been imported.

```python
menu = "\nChoose the backend:\n1. TensorFlow\n2. PyTorch\nYour choice: "
while True:
  choice = input(menu).strip()
  if choice == "1":
    backend = "tensorflow"
    break
  elif choice == "2":
    backend = "torch"
    break
```

```python
import os
os.environ["KERAS_BACKEND"] = backend
```

```python
import sys
print( f"Python {sys.version}" )

import numpy as np
print( f"NumPy {np.__version__}" )

import matplotlib as mpl
print( f"Matplotlib {mpl.__version__}" )
```

```python
if backend == 'tensorflow':
  import tensorflow as tf
  print( f"TensorFlow {tf.__version__}" )
  print( f"|- tf.keras.backend.image_data_format() = {tf.keras.backend.image_data_format()}" )

  # Count the number of GPUs as detected by tensorflow
  gpus = tf.config.list_physical_devices('GPU')
  print( f"|- TensorFlow detected { len(gpus) } GPU(s):" )
  for i, gpu in enumerate(gpus):
    print( f"   |- GPU {i}: Name = {gpu.name} , Type = {gpu.device_type}" )

elif backend == 'torch':
  import torch
  print( f"PyTorch {torch.__version__}" )

  # Count the number of GPUs as detected by pytorch
  if torch.cuda.is_available():
    num_gpus = torch.cuda.device_count()
    print(f"|- PyTorch detected  {num_gpus} GPU(s):")
    for i in range(num_gpus):
        print(f"   |- GPU {i}: {torch.cuda.get_device_name(i)}")
  else:
    print("|- PyTorch detected no GPU. Running on CPU.")

import keras
print( f"\nKeras {keras.__version__}" )
```
